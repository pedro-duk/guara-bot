```python
# Convert every PNG in the sibling "png" folder into the sibling "webp" folder.

from pathlib import Path

from PIL import Image


try:
    script_dir = Path(__file__).resolve().parent
except NameError:
    script_dir = Path.cwd().resolve()
png_folder = script_dir / "png"
webp_folder = script_dir / "webp"
webp_folder.mkdir(parents=True, exist_ok=True)

for image_path in png_folder.iterdir():
    if not image_path.is_file() or image_path.suffix.lower() != ".png":
        continue

    with Image.open(image_path) as image:
        mode = "RGBA" if "A" in image.getbands() or "transparency" in image.info else "RGB"
        output_path = webp_folder / f"{image_path.stem}.webp"
        image.convert(mode).save(output_path, format="WEBP")
```
